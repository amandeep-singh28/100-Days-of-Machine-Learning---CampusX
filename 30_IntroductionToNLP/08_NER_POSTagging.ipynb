```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
# sm - small model
```

- NER models are not perfect, it can make mistakes

- When we write doc = nlp(text), spacy sends the text through an NLP pipeline. For example:- Raw test -> Text Processing -> POS tagging -> NER, etc

```python
text = "Amandeep lives in Haridwar and studies at Lovely Professional University"
text2 = "Sundar Pichai visited New York on 15 January 2025 and met employees of Google."
doc = nlp(text)
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_}")

doc2 = nlp(text2)
for ent in doc2.ents:
    print(f"{ent.text} -> {ent.label_}")
```

```python
for token in doc:
    print(token.text)
```

```python
for token in doc:
    print(token.text, token.pos_)
```

```python
text = "The smart student studies NLP"
doc = nlp(text)
for token in doc:
    print(f"{token.text} -> {token.pos_}")
```
