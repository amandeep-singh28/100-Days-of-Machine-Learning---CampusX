```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love NLP",
    "NLP is amazing",
    "I love AI"
]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
```

- CountVectorizer stores the vocabulary in sorted order and also it converts the text into lowercase too
- CountVectorizer also ignores those words which are of single character only

```python
print(X.toarray())
```

```python
print(vectorizer.vocabulary_)
```

- Here we are using transform(), so we are just applying what we have learned before

```python
new_doc = ["I love amazing NLP"]

print(vectorizer.transform(new_doc).toarray())
```

```python
new_doc = ["I love machine learning"]
print(vectorizer.transform(new_doc).toarray())
```
