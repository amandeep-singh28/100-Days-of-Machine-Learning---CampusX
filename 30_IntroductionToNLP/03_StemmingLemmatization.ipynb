#### Stemming

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
print(ps.stem("playing"))
print(ps.stem("studies"))
print(ps.stem("running"))
print(ps.stem("history"))
```

```python
from nltk.tokenize import word_tokenize
sentence = "I am playing cricket and studying NLP"
words = word_tokenize(sentence)
print(words)

for i in words:
    print(ps.stem(i))
```

#### Lemmatization

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("studies"))
print(lemmatizer.lemmatize("running"))
print(lemmatizer.lemmatize("better"))
```

```python
from nltk.tokenize import word_tokenize
sentence = "I am playing cricket and studying NLP"
words = word_tokenize(sentence)
print(words)

for i in words:
    print(lemmatizer.lemmatize(i))
```

```python
print(lemmatizer.lemmatize("running"))
```

```python
print(lemmatizer.lemmatize("better"))
```

- In theory, we have discussed that if we lemmatize 'running' than it would be 'run', but here nltk is thinking that its a noun. So we need to tell nltk that its a verb
- verb: v, adjective: a, noun: n

```python
print(lemmatizer.lemmatize("running", pos = 'v'))
```

```python
print(lemmatizer.lemmatize("better", pos='a'))
```

```python
from nltk.tokenize import word_tokenize
sentence = "I am playing cricket and studying NLP"
words = word_tokenize(sentence)
print(words)

for i in words:
    print(lemmatizer.lemmatize(i, pos = 'v'))
```

```python
print(lemmatizer.lemmatize("studies"))
print(lemmatizer.lemmatize("studying"))
print(lemmatizer.lemmatize("studying", pos='v'))
```
