### Text Preprocessing

```python
text = text.lower()
print(text)
```

```python
# punctuation removal
import re 
text = re.sub(r'[^\w\s]', '', text)
print(text)
```

- [^\w\s] -> It means remove everything except letters, numbers and space
- re -> regular expression library which is used for Searching text, Finding patterns, Replacing text and Validating text
- re.sub(pattern, replacement, text)

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
print(stop_words[:5])
```

```python
# removing stopwords
words = text.split()
filtered_words = []
for i in words:
    if i not in stop_words:
        filtered_words.append(i)
print(filtered_words)
```
